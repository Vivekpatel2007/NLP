Q1. Construct a deterministic finite automaton (DFA) that recognizes valid simplified English
words. An English word, for this problem, is defined as a string that:

● Starts with an lowercase letter

● Followed by zero or more lowercase letters

That is, the string must:
● Not start with a digit or punctuation

● Not contain uppercase letters after the first character

● Not contain spaces, digits, or special characters

Output Format:
If a string is accepted, output Accepted else Not Accepted.

● Accepted: cat, dog, a, zebra

Not Accepted: dog1, 1dog, DogHouse, Dog_house, cats (starts with a space)
For visualizing the DFA, you can use https://pypi.org/project/visual-automata/

```python
from automata.fa.dfa import DFA
from visual_automata.fa.dfa import VisualDFA
```

```python
import string

lowercase = set(string.ascii_lowercase)
uppercase =set(string.ascii_uppercase)
alphabets=set(lowercase.union(uppercase))
all_letters=set(string.printable)
```

```python
dfa = VisualDFA(
    states={"q0", "q1", "qd"},
    input_symbols= all_letters,
    transitions={
        "q0": { ch : ("q1" if ch in lowercase else "qd") for ch in all_letters},
        "q1": { ch : ("q1" if ch in lowercase else "qd") for ch in all_letters},
        "qd": { ch : ("qd") for ch in all_letters}
    },
    initial_state="q0",
    final_states={"q1"},
)
```

```python
dfa.table
```

```python
test=["cat","dog","a","dog1", "1dog", "Doghouse", "Dog_house"]
for i in test:
  print(dfa.input_check(i))
```
